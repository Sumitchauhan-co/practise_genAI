```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
texts = "Project Nebula Overview Project Nebula is an internal initiative initiated in Q3 2025 to overhaul the cloud infrastructure. The primary goal is to reduce latency by 45% and migrate all legacy microservices to a containerized Kubernetes architecture.  Lead Architect: Dr. Elena Vance. Budget: $4.2 Million. Deadline: December 2026."
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=50, chunk_overlap=20)
```

```python
chunks = splitter.split_text(texts)
chunks
```

Document Text Loader


```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../sample.txt")
texts = loader.load()
texts
```

```python
pdfLoader = PyPDFLoader("../sample.pdf")
docs = pdfLoader.load()
docs
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=25)
```

```python
text_chunks = splitter.split_documents(texts)
text_chunks
```

```python
pdf_chunks = splitter.split_documents(docs)
pdf_chunks
```
