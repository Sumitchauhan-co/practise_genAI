```python
from pydantic import BaseModel
```

```python
class User(BaseModel):
    email: str
    name: str


user = User(email="one@one.com", name="one")
user
```

```python
from pydantic import Field
from typing import Literal
import uuid
```

```python
class Todo(BaseModel):
    id: str = uuid.uuid4()
    title: str = Field(max_length=50)
    decsription: str = Field(min_length=10)
    status: Literal["pending", "completed"] = "pending"
```

```python
todo = Todo(
    title="Diet plan",
    decsription="6-6-6-6 diet plan to gain 100gm of protein daily",
    status="pending",
)
todo
```

```python
class Product(BaseModel):
    id: str = uuid.uuid4()
    name: str
    price: int = Field(gt=0)


product = Product(name="Product_eyowslk", price=100)
```

```python
class Order(BaseModel):
    id: str = uuid.uuid4()
    product: Product


order = Order(product=product)
```

```python
productId = order.product.id
productId
```

```python
from typing import List
```

```python
class Cart(BaseModel):
    id: str = uuid.uuid4()
    items: List[Product]


cart = Cart(items=[product])
cart
```

Annotator


```python
from typing import Annotated
```

```python
commonValidator = Annotated[
    List[Product], "This is common validator", Field(default=[])
]
```

```python
class Cart_(BaseModel):
    id: str = uuid.uuid4()
    items: commonValidator
```

```python
class Order(BaseModel):
    id: str = uuid.uuid4()
    product: commonValidator


ord = Order(product=[product])
ord
```
