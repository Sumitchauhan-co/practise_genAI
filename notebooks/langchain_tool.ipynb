```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_two_numbers(a: int, b: int):
    """
    It will return the sum of two integer numbers
    Args:
      a: First Number
      b: Second Number
    """
    return a + b


@tool
def mutlipy_two_numbers(a: int, b: int):
    """
    It will return the multiplication of two integer numbers
    Args:
      a: First Number
      b: Second Number
    """
    return a * b
```

```python
add_two_numbers.invoke({"a": 12, "b": 16})
```

```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
```

```python
llm = ChatOllama(model="minimax-m3:cloud")
```

```python
agent = create_agent(
    model=llm,
    tools=[add_two_numbers, mutlipy_two_numbers],
    system_prompt="You are a Math teacher and always use tools for calculation.",
)
```

```python
res = agent.invoke({"messages": [{"role": "user", "content": "What is 3 + 7 * 5?"}]})
```
