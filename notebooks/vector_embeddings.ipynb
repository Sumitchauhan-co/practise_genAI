```python
from dotenv import load_dotenv

load_dotenv()
```

```python
documents = [
    "The early morning fog slowly cleared from the valley.",
    "Reading books opens up new worlds of imagination.",
    "Regular exercise helps maintain physical and mental health.",
    "Learning a new skill requires patience and practice.",
    "The automated system processed the data in seconds.",
    "A warm cup of tea is perfect for rainy afternoons.",
    "Innovation drives progress in the modern tech industry.",
    "Nature sounds can greatly reduce daily stress levels.",
    "Effective communication is the key to successful teamwork.",
    "They decided to take a spontaneous road trip over the weekend.",
]
```

```python
from langchain_ollama import OllamaEmbeddings
```

```python
embeddings = OllamaEmbeddings(model="qwen3-embedding")
```

```python
vector = embeddings.embed_documents(documents)
len(vector[0])
```

Store Vectore Embeddings in Vector DB


```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=documents, embedding=embeddings, persist_directory="../vector.db"
)
```

```python
query = "?"
result = vector_store.similarity_search(query=query, k=2)
result
```
